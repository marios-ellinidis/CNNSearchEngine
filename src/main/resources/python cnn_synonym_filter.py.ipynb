```python
import pandas as pd

from nltk.corpus import stopwords
import string
from tqdm import tqdm  # Import tqdm for the progress bar
import nltk
nltk.download('stopwords')

# Load stopwords and define fields to extract from
stop_words = set(stopwords.words('english'))
fields_to_use = ['Headline', 'Description', 'Keywords', 'Second headline', 'Article text']

# Load articles
articles_df = pd.read_csv('CNN_Articels_clean.csv')  # replace with your actual file path

# Collect unique words from the selected fields
def extract_words(text):
    if pd.isna(text):
        return []
    words = text.lower().translate(str.maketrans('', '', string.punctuation)).split()
    return [w for w in words if w not in stop_words and w.isalpha()]

unique_words = set()

# Use tqdm to show the progress of applying the word extraction across all fields
for field in fields_to_use:
    articles_df[field] = articles_df[field].astype(str)  # make sure it's all strings
    for text in tqdm(articles_df[field], desc=f"Processing {field}", position=0, leave=True):  # Add progress bar
        unique_words.update(extract_words(text))

print(f"Total unique non-stop words from articles: {len(unique_words)}")

# Load WordNetSynonyms.csv
synonyms_df = pd.read_csv('WordnetSynonyms.csv', usecols=['Word', 'Synonyms'])

# Normalize the Word column to lowercase for comparison
synonyms_df['Word'] = synonyms_df['Word'].str.lower()

# Filter synonyms only for relevant words
filtered_synonyms = synonyms_df[synonyms_df['Word'].isin(unique_words)]

# Save to a new CSV
filtered_synonyms.to_csv('FilteredSynonyms.csv', index=False)

print(f"Filtered synonyms saved to 'FilteredSynonyms.csv' with {len(filtered_synonyms)} entries.")
```
